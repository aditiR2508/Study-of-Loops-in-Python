# While Loop

```python
# While Loop---> used when the number of iterations is not known.
### Syntax: while condition:
#### statement(s)
```

```python
# To print i as long as i is less than 6:
i=1
while i <= 10:
  print(i)
  i += 1
```

```python
# To print numbers from 1 to n---> take user input for n
i = 1
n= int(input("Enter an end value: "))
while i <= n:
  print(i)
  i += 3
```

```python
# Factorial of a number
a = int(input("Enter a number: "))
f = 1
while a>0:
  f *= a
  a -= 1

print("Factorial: ", f)
```

```python
# Fibonacci Series
x=0
y=1
n=int(input("Enter a number: "))
while x <= n:
  print(x, end=" ")
  x= y
  y = x+y
```

```python
#Reverse of a number
a = int(input("Enter a number: "))
n=0
while a>0:
  x=a%10
  n=n*10 +x
  a//=10   ##floor division

print("Reversed Number= ", n)
```

```python
#palindrome number
x = int(input("Enter a number: "))
original_x = x ##keeping a copy of the original number as 'x' will be reduced to 0 during the loop
rev_x = 0

if x<0:
  is_palindrome = False  ##checking for negative number first
else:
  while x>0:
    digit = x % 10            ## for getting the last digit
    rev_x = (rev_x*10) + digit ## move digits to the left and adds new one
    x = x // 10                ## remove the last digit from the number
  is_palindrome = (original_x == rev_x)

if is_palindrome:
  print("Number provided is Palindrome!")
else:
  print("Number provided is not Palindrome.")
```

```python
# To check Palindromes----> usage of while, if and break statements
a = str(input("Enter a string: "))
i = 0
j = len(a)-1
is_palindrome = True

while i<j:
  if a[i] != a[j]:
    is_palindrome = False
    break
  i += 1
  j -= 1

if is_palindrome:
  print("Given string is a palindrome.")
else:
  print("Given string is not a palindrome.")
```

```python
# To count digits in a number
n = int(input("Enter a number: "))
a = 0
while(n>0):
  a+=1
  n//=10

print("The number of digits: ", a)
```

## Break Statement
### --> With this statement, we can stop the loop even if the while condition is true.

```python
# To exit the loop when i is 3:
i = 1
n=int(input("Enter a number: "))
while i < n:
  print(i)
  if i==3:
    break
  i+=1
```

```python
# Search an element in a list--> take user input for the element you want to search
marks=[57, 78, 90, 85, 79]
a = int(input("Enter marks to search: "))
i=0
while i<len(marks):
  if marks[i]==a:
    print("Element was found at position", i)
    break
  i+=1
else:
  print("Element was not found.")
```

# For Loop

```python
#Problem Statement 1: python program to accept three digits and print all possible combinations
a = int(input("Enter first digit: "))
b = int(input("Enter second digit: "))
c = int(input("Enter third digit: "))

d = [a,b,c]

for i in range(0,3):
  for j in range(0,3):
    for k in range(0,3):
      if (d[i]!=d[j] and d[j]!=d[k] and d[k]!=d[i]):
        print(d[i],d[j],d[k])
```

```python
# Armstrong Number(153 370 371 407 123)---> an 'n' digit integer where the sum of its digits, each raised to the 'n-th' power, equals the original number
x = int(input("Enter a number: "))
total = 0

for digit in str(x):    ## int datatype is not iterable while str is iterable
  total += int(digit)**len(str(x))

if total == x:
  print("Provided number is an Armstrong Number!")
else:
  print("Try Again! Not an Armstrong Number!!")
```

```python
# To print all prime numbers in Range
for x in range(2,100):
  for i in range(2,x):
    if x%i == 0:
      break
  else:
    print(x)
```

```python
# To print an inverted pyramid
a = 5
for i in range(a):     ## 0, 1, 2, 3, 4
  print(i*' ', (a-i)*' $')
```

```python
# to print a pyramid(normal)
a = 5
for i in range(a):
  print((a-i)*' ', (i+1)*' $')
```

```python
# Right angled triangle
a = 5
for i in range(a):
  print((i+1)*' $')
```

```python
# Left Angled triangle
a = 5
for i in range(a):
  print(' '*(a-i), i * '$')
```

```python
# Inverted Right angled triangle
a = 5
for i in range(a):
  print((a-i)*' $')
```

```python
# Inverted Left angled triangle
a = 5
for i in range(a,0,-1):
  print(' '*(a-i), '$'*i)
```

```python
# Square
a = 5
for i in range(a):
  print((a+1)*' $')
```

```python
# Rhombus
a = 5
for i in range(a):
  print(' '*(a+i), a*' $')
```

```python
# Diamond
a = 5
for i in range(1, a + 1):
    print(' ' * (a - i) , ' $' * i)      ## Upper part

for i in range(a - 1, 0, -1):
    print(' ' * (a - i) , ' $' * i)       ## Lower part
```
